## Teen Smartphone Addiction Analysis

---



`About Data`:

This dataset focuses on understanding the patterns and effects of smartphone usage among teenagers, with a special focus on:

Addiction levels

Academic performance

Time distribution (social media, gaming, education)

Mental and physical health proxies

It enables researchers, educators, psychologists, and data scientists to explore and model behavioral trends, and build classification or regression models to predict addiction levels, performance, or placement outcomes.

`Purpose of Analysis`:
- To explore how smartphone addiction among teenagers varies across gender, grade, and behavioral traits, and to uncover how this addiction correlates with mental health factors such as anxiety, depression, self-esteem, and family communication.

- The goal is to identify which groups are more prone to addiction, how it impacts their daily app usage, time spent on social media, and to inform future interventions or awareness programs targeting teen smartphone overuse.

```python
# checking outliers thorugh boxplot

for i , col in enumerate(df.drop(columns = ["Age" , "Gender" ,"School_Grade" , "Phone_Usage_Purpose"])):
  plt.figure(figsize=(6, 4))
  sns.boxplot(data=df, x=col)
  plt.title(f'Boxplot of {col}')
  plt.xticks(rotation=30)
  plt.tight_layout()
  plt.show()
```

```python
# Select numeric columns only
numeric_cols = df.select_dtypes(include=['number'])

# Dictionary to store outliers
outliers = {}

for col in numeric_cols.columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Find outliers
    outlier_values = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]

     #store columns with outliers
    if outlier_values.count()!=0:

      # Store results
      outliers[col] = {
          # "Q1": Q1,
          # "Q3": Q3,
          # "IQR": IQR,
          # "Lower Bound": lower_bound,
          # "Upper Bound": upper_bound,
          "Outlier Count": outlier_values.count(),
          "Outlier Values": list(outlier_values)
      }

# Example: show outliers in 'Daily_Usage_Hours'
print(outliers)

```

- `Outliers Present`
- According to context these outliers can be outliers but leaving them according to domain knowldge because teens students having 12 hours screen time can be normal . We will handle it according to future needs

## Insights


High addiction is linked to Poor sleep , low self-esteem, high anxiety, and low family communication and high depression.

Social media is the leading cause of phone usage among teens.

Gender and grade have moderate influence on addiction patterns.(mostly no impacts)


---

### 📱 **Addiction and Daily Usage**

* **Males from 7th grade** have **high addiction levels** and spend **more daily hours** on their phones.
* **All genders** with **high addiction** spend **more time on social media**.

---

### 📱 **Social Media Usage**

* **Males from 12th grade** with **low addiction** spend **more time on social media** than females and others with low addiction.

---

### 📲 **App Usage**

* **Females from 11th grade** with **low addiction** use **more apps daily** than males and others with low addiction.
* **Others from 11th grade** with **high addiction** use **more apps** than both males and females with high addiction.

---

### 🎮 **Gaming Time**

* **Females from 11th grade** with **low addiction**, and **females from 7th grade** with **medium addiction**, spend **more time on gaming**.
* **Males from 9th grade** with **high addiction** also spend **more time on gaming**.

---

### 📞 **Phone Checks Per Day**

* **Females from 9th grade** and **others from 7th grade** with **low addiction** check their phones **more frequently**.
* **Females from 7th grade** with **medium addiction** also have **frequent phone checks**.
* **Males from 8th grade** with **high addiction** have **high phone checks**, though values are mostly similar across high addiction.

---

### 😴 **Sleep Hours**

* **Males from 9th grade** with **low addiction** have **more sleep hours** than females and others.
* Across all genders, those with **high addiction** have **similar (lower) sleep hours**.

---

### 🧠 Final Summary:

* **High addiction** is consistently linked with **increased screen behaviors** (social media, gaming, phone checks) across genders and grades.
* **Some low addiction groups**, especially **females and males in senior grades**, still exhibit high usage in specific areas like social media and apps.
* **Sleep decreases** as addiction level increases, while **middle schoolers** (7th–9th grade) are the most affected across multiple behavior indicators.

---


## 📊 Final Insights Summary

1. **Addiction and Gender**:
   - Males from 7th grade show high addiction levels.
   - Surprisingly, 12th-grade males with *low addiction* still spend more hours on mobile than others.

2. **App Usage Patterns**:
   - Females from 11th grade with low addiction levels use more apps daily compared to other groups.
   - “Other” gender from 11th with high addiction uses more apps than males/females.

3. **General Trends**:
   - All genders with high addiction levels also spend more time on social media.
   - Students with high anxiety or depression levels tend to have higher smartphone addiction.

4. **Behavioral Patterns**:
   - Poor family communication is often linked with high addiction and low self-esteem.
   - Students who use phones mostly for **social media** show higher addiction compared to those using them for **education** or **calls**.
